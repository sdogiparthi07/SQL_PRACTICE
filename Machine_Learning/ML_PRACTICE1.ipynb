How Linear Regression works?
Linear Regression is one of the simplest machine learning algorithms. It tries to find the best straight line that describes the relationship between an input variable x and an output variable y.

WHERE LINEAR REGRESSION IS USED?

Linear regression is used whenever you want to predict a continuous numerical value based on one or more input variables.
It is widely used in machine learning, data science, finance, healthcare, business, and engineering.

Here's a visualization of what linear regression does—it fits the best straight line through data points to make predictions:
Example 1: House Price Prediction 
House Size (sq ft)	Price ($)
1000	200,000
1500	250,000
2000	300,000
2500	350,000

Suppose you are a real estate agent. You have information about houses that have already been sold.
You notice that larger houses usually cost more.

Instead of manually guessing the price of every new house, you train a linear regression model using the past data.The model learns a relationship like:
Price=0.1×Size+100
Now, a customer asks:
"How much would a 1800 sq ft house cost?"
The model substitutes 1800 into the equation and predicts about $280,000.
So, linear regression uses previous house sales to estimate the price of a new house.

When should you use Linear Regression?

The output is a number (price, salary, temperature, marks, etc.).
The relationship between inputs and output is approximately linear (a straight-line trend).

1. Residual : A residual is the difference between the actual value and the value predicted by the regression model. It tells us how much the model's prediction is different from the real value.
    Formula  :    Residual = Actual Value − Predicted Value ;   e=y−y^
	​Where: y = actual output value; y^ = predicted output value;  e = residual/error
Example :
Suppose we predict house prices. Actual price: y=$300000
Model predicted price:y^ =$280000
Residual:
        e=300000−280000
        e=20000 .The model underestimated the house price by $20,000.
Why Residuals are Important?
Residuals help us understand:
How accurate the model is.
Whether the model is making large mistakes.
Whether there are unusual data points (outliers).
Whether a linear model is suitable for the data.
A good regression model should have small residuals.

OLS (Ordinary Least Squares) : Ordinary Least Squares (OLS) is a statistical method used to find the best-fitting line in linear regression.
The objective of OLS is: Find the line that minimizes the difference between actual values and predicted values.
The regression equation: y=mx+c
OLS finds the best values of:

m → slope
c → intercept
Why "Least Squares"? Because OLS minimizes the sum of squared residuals.
Formula: SSE=∑(y−y^)2
Where: SSE = Sum of Squared Errors
y  = actual value
y^ = predicted value

OLS Working
Take the dataset.
Try different regression lines.
Calculate residuals.
Square residuals.
Add squared errors.
Select the line with minimum error.

Advantages of OLS
Simple and easy to interpret.
Gives an exact solution for linear regression.
Works well for small and medium datasets.

Limitations of OLS
Slow for very large datasets.
Sensitive to outliers.
Requires assumptions about the data.



3. Gradient Descent : Gradient Descent is an optimization algorithm used to find the best values of model parameters by reducing the error step by step. It is commonly used when datasets are large.
Basic Idea
Imagine standing on a mountain and trying to reach the lowest point.

The slope tells you which direction to move.
The learning rate tells you how big a step to take.
You keep moving until you reach the minimum point.

In regression:
Mountain height = error/cost function.
Lowest point = minimum error.
Cost Function

Gradient Descent tries to minimize:  J= 1/n ∑(y−y^)2
Where: J = cost/error  n = number of data points

Parameter Update Formula
Updating slope: m=m−α*∂m/∂J
Updating intercept: c=c−α ∂c/ ∂J
	​Where:
m = slope
c = intercept
α = learning rate
∂m/∂J = gradient

Gradient Descent Applications used in:
Linear regression
Logistic regression
Neural networks
Deep learning


example : Real-Life Example - House Price Prediction. A real estate company wants to predict house prices.
Input: X=House Size
Output: Y=Price
Initial model: Price=0(Size)+0
Predictions are wrong.
Gradient Descent adjusts:
m → price increase per square foot
c → starting price
After many iterations: Price=200(Size)+50000
Now, House size: 1000 sq.ft
Prediction: Price=200(1000)+50000 =250000
The model predicts: $250,000



R² (Coefficient of Determination) : R² measures how well the regression model explains the variation in the output data.

It answers: How much of the change in Y is explained by X?
Formula
         R2 =1−SSres/ SStot
	​Where: Residual Sum of Squares: SSres =∑(y−y^)2
           Total Sum of Squares = SStot =∑(y-y)2
    Where: yˉ=average of actual values

Adjusted R² : Adjusted R² is a modified version of R² that considers the number of features used in the model. It solves a problem with R²:
R² always increases when new features are added, even if they are not useful.
Formula : AdjustedR^2=1−(1−R2)n−1/n−p-1
	​Where: n = number of observations
           p = number of independent variables/features
Understand the Formula
Adjusted R² formula: Adjusted R2=1−(1−R2)(n−1/n−p−1)
Where:
R2= R-squared value
n = number of observations (data points)
p = number of independent variables (features)
Step 2: Identify Values
From our example:
Data points:
Students: A, B, C, D, E
So:n=5
Independent variable:
We used only: X=Study Hours
So: p=1     R² value:R
Step 3: Substitute Values
Formula: Adjusted R2 =1−(1−0.875)(5−1/5-1-1)
First calculate:1−R2
                1−0.875=0.125
Now calculate the fraction:    n−1 / n−p−1
                            =  5−1/ 5−1−1
                            =   4/3
Now:
Adjusted R2=1−(0.125)(4/3)
           =1−(0.125)(1.333)
           =1−0.1666
Adjusted R2=0.833
After considering the number of features used, the model explains: 83.3% of the variation.
The value decreased because Adjusted R² applies a penalty for using predictors.

Difference Between R² and Adjusted R²
R²	                                  Adjusted R²
Measures model accuracy	              Measures model accuracy with penalty
Always increases with more features	  May decrease with unnecessary features
Used for simple regression	          Used for multiple regression
Can overestimate performance	      Gives a more realistic score
