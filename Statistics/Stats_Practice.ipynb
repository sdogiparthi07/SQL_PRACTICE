Statistics: is the science of collecting, organizing, analyzing, interpreting, and presenting data to make decisions.
1.Descriptive Statistics: Summarizes and describes data.
Examples: tables, graphs, averages, percentages.
2.Inferential Statistics: Uses sample data to make predictions or conclusions about a population.
Examples: hypothesis testing, confidence intervals.
BASIC STATISTICS TERMS
1.Population: The complete group being studied.
Example: All students in a school.
2.Sample: A smaller group selected from a population.
Example: 100 students chosen from the school.
3.Variable: A characteristic that can take different values.
Examples: age, height, income, test scores.
4.Data: Collected observations or measurements.


TYPES OF DATA
1.CATEGORICAL(QUALITATIVE): Describes qualities or categories.
Examples:Gender,Color, Type of vehicle
2.NUMERICAL (QUANTITATIVE): Represents numbers.
Examples:Height, Weight, Salary
Levels of Measurement in Statistics
1.Nominal Scale: Data is grouped into categories with no order or ranking.
Example: Gender, blood type, eye color
2.Ordinal – Data has categories with a meaningful order or ranking, but the differences between ranks are not equal.
Example: Satisfaction level (poor, good, excellent), class rank
3.Interval – Data has ordered values with equal differences between them, but there is no true zero point.
Example: IQ scores, temperature in Celsius, calendar years
4.Ratio – Data has ordered values, equal intervals, and a true zero point, allowing all mathematical calculations.
Example: Height, weight, age, income
TYPES OF DATA IN NUMERICAL : 
Discrete Data: It has a particular fixed value and can be counted.
Continuous Data: It is not fixed but has a range of data and can be measured. (height, temperature)


MEASURE OF CENTRAL TENDENCY
 MEAN: The mean can be calculated by summing all values present in the sample divided by total number of values present in the sample or population.
FORMULA: MEAN = sum of values / number of values
#example A teacher wants to find the average test score of 5 students.
Test scores: 70, 80, 85, 90, 95
Mean = (70 + 80 + 85 + 90 + 95) ÷ 5
Mean = 420 ÷ 5
Mean = 84
2.Median: The median is the middle of a dataset when arranged from lowest to highest or highest to lowest in order to find the median, the data must be sorted. For an odd number of data points the median is the middle value and for an even number of data points median is the average of the two middle values.
Example: A company looks at the salaries of 5 employees. Salaries: $30,000, $40,000, $50,000, $60,000, $100,000
The middle value is $50,000. The median salary is $50,000.
(Median is useful because the very high salary of $100,000 does not affect it much.)


3. Mode: The most frequently occurring value in the Sample or Population is called Mode.
A clothing store records the most common shoe size sold.
Shoe sizes sold: 7, 8, 8, 9, 10, 8, 11
The size that appears most often is 8. The mode is shoe size 8.

	Range: Range is the difference between the maximum and minimum values of the Sample.

                                           Formula:σ^2 = (Σ(X-μ)^2)/n

Measure of Shape
1.SKEWNESS: is the measure of asymmetry of probability distribution about its mean.
Types of Skewed data
•Positive Skew (Right): Mean > Median
•Negative Skew (Left): Mean < Median
•Symmetrical: Mean = Median
Symmetrical means that two sides of something are the same or balanced when divided along a central line or point.
Example:
•A butterfly has symmetrical wings — the left and right sides look the same. 
•A human face is approximately symmetrical.
Asymmetrical means that two sides are not the same or not balanced. The parts are different in size, shape, or arrangement.
Example:
•A tree branch pattern is often asymmetrical. 
•A painting with objects placed unevenly can be asymmetrical.

Interquartile Range (IQR): The range between the first quartile (Q1) and the third quartile (Q3). It is less sensitive to extreme values than the range. To compute IQR, calculate the values of the first and third quartile by arranging the data in ascending order. Then, calculate the mean of each half of the dataset.
                            Formula:IQR=Q_3-Q_1
Quartiles: Quartiles divides the dataset into four equal parts:
Q1 (First Quartile): Median of the lower 50% of the dataset (25th percentile).
Q2 (Second Quartile / Median): Median of the entire dataset (50th percentile).
Q3 (Third Quartile): Median of the upper 50% of the dataset (75th percentile).


•Probability Distribution Function: describes how the probabilities of a random variable are distributed among its possible values. It tells us the likelihood that a random variable will take on a particular value or fall within a certain range. The total probability of all possible outcomes is always 1. 
There are two types of probability distribution functions:
Probability Mass Function (PMF) – For Discrete Random Variables
•Used for discrete data (countable values). 
•Gives the probability of each individual value. 
•The sum of all probabilities is 1. 
Example: Rolling a fair die
Outcome (X)	1	2	3	4	5	6
P(X)	1/6	1/6	1/6	1/6	1/6	1/6
Here, the probability distribution shows that each outcome has an equal probability of 1/6.
Probability Density Function (PDF) – For Continuous Random Variables
•Used for continuous data (measurable values). 
•The probability of one exact value is 0. 
•It shows how probabilities spread over a range but not the chance of one exact value since values can be infinite.
•The probability is found by calculating the area under the curve between two values. 
•The total area under the curve is 1. 
Example: Student heights are normally distributed. The probability that a student's height is between 160 cm and 170 cm is the area under the curve between these two heights.


Types of Probability Distributions
Probability distributions can be divided into two main types based on the nature of the random variables: discrete and continuous.
Discrete: A discrete distribution is used when the random variable can take on countable, specific values. 
Example : when predicting the number of products, a customer buys in a single order the possible outcomes are whole numbers like 0, 1, 2, 3, etc. You can't buy 2.5 products so this is a discrete random variable.
Common Probability Distributions
•Binomial Distribution: The binomial distribution calculates the chance of getting a certain number of successes in a fixed number of trials.
•Example: flipping a coin 10 times and counting heads.
•Nuber of trials: 10
•Two outcomes per trial: heads (success) or tails (failure)
•Probability of success (heads): 0.5
•Shows likelihood of getting 0 to 10 heads
2. Bernoulli Distribution: The Bernoulli distribution describes experiments with only one trial and two possible outcomes: success or failure. It’s the simplest probability distribution.
 Example: flipping a coin once and checking if it lands on heads.
•One trial only
•Two outcomes: heads (success) or tails (failure)
•Probability of success: 0.5
•Graph shows two bars representing success (1) and failure (0) with equal probabilities
Continuous: is used when a random variable can take any value within a given range, including decimal values. Unlike discrete data, continuous data is measured rather than counted.
Example: Height, weight, temperature, or the amount of money a customer spends.
•Normal Distribution: A bell-shaped distribution in which most values cluster around the mean. 
•Uniform Distribution: Every value within a given range has an equal probability of occurring. 
•Exponential Distribution: Models the time between independent events that occur at a constant average rate.
